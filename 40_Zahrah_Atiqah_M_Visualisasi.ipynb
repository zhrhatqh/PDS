```python
#Data
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Juni', 'Juli']
penjualan = [50,100,150,200,250,300,350]
plt.figure(figsize=(8, 4))

#1. Buat grafik batang
sns.barplot(x=bulan, y=penjualan, palette='muted')

#2. Tambahkan garis tren (menghubungkan puncak batang)
plt.plot(
    range(len(bulan)),
    penjualan,
    color='pink',
    marker='o',
    linestyle='-',
    linewidth=2,
    label='Garis Tren'
)

#3. Kustomisasi Tampilan
plt.title('Penjualan per bulan dengan garis tren', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Jumlah Penjualan')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv ("Dataset Pandas.csv")
df

df['total'] = df[['jan', 'feb', 'mar', 'apr']].sum(axis=1)
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8,5))
df_sorted = df.sort_values('total', ascending=False)
sns.barplot(data=df_sorted, x='nama', y='total', palette='viridis')

plt.title('Total Penjualan per Orang', fontsize=12)
plt.xlabel('Nama Karyawan')
plt.ylabel('Total Penjualan')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Pie chart
plt.figure(figsize=(5, 5)) # Ukuran persegi agar pie chart tidak gepeng

# Hitung total penjualan per bulan
total_per_bulan = df[['jan', 'feb', 'mar', 'apr']].sum()

plt.pie(total_per_bulan, labels=total_per_bulan.index, autopct='%1.1f%%',
        startangle=140, colors=sns.color_palette('colorblind'))

plt.title('Proporsi Penjualan per Bulan', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Ubah format data menjadi 'long'
df_melted = df.melt(
    id_vars='nama',
    value_vars=['jan', 'mar'],
    var_name='Bulan',
    value_name='Penjualan'
)

# Buat Line Chart
fig = px.line(
    df_melted,
    x='nama',
    y='Penjualan',
    color='Bulan',
    symbol='Bulan',
    symbol_sequence=['circle', 'triangle-up'],
    color_discrete_sequence=['pink', 'yellow'],
    markers=True,
    title='Perbandingan Penjualan: Januari vs Maret',
    labels={
        'nama': 'Nama Karyawan',
        'Penjualan': 'Jumlah Penjualan'}
)

# Atur ukuran
fig.update_layout(width=800, height=500)

# Tampilkan grafik
fig.show()
```
