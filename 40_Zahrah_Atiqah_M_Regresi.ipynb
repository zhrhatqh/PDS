```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

data = {
    'Partisipan': np.arange (1, 11),
    'PMakan': [22, 34, 20, 50, 41, 32, 60, 77, 82, 90],
    'PTranspor': [20, 30, 40, 50,10, 60, 90, 88, 92, 38],
    'BB': [45, 43, 55, 49, 40, 60, 50, 39, 88, 67]
}

df = pd.DataFrame (data)

x = df[['PMakan', 'PTranspor']]
y = df['BB']

x = sm.add_constant(x)
model = sm.OLS(y, x)
results = model.fit()

print(results.summary())
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Dataset Fiskal_Pekanbaru.csv')
df.head()

# VARIABEL INDEPENDEN
X = df[["bm_pendidikan", "bm_ekonomi"]]

# VARIABEL DEPENDEN
y = df["KEMISKINAN"]

# MENAMBAHKAN INTERCEPT
X_model = sm.add_constant(X)

# ESTIMASI MODEL OLS
model = sm.OLS(y, X_model)
results = model.fit()

# MENAMPILKAN HASIL REGRESI
print("\n")
print("=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA - KOTA PEKANBARU")
print("=" * 70)
print(results.summary())

# MENGAMBIL KOEFISIEN REGRESI
intercept = results.params["const"]
coef_pendidikan = results.params["bm_pendidikan"]
coef_ekonomi = results.params["bm_ekonomi"]

r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# MEMBUAT PERSAMAAN REGRESI
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_pendidikan:.3f}) X1"
    f" + ({coef_ekonomi:.3f}) X2"
)

print("\n")
print("=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# MENGHITUNG NILAI PREDIKSI
df["Prediksi"] = results.predict(X_model)
print("\n")
print("=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)

print(
    df[["Tahun","bm_pendidikan","bm_ekonomi","KEMISKINAN"
    ]].round(3)
)

# MEMBUAT GRID UNTUK BIDANG REGRESI
x1 = np.linspace(
    df["bm_pendidikan"].min(),
    df["bm_pendidikan"].max(),
    40
)

x2 = np.linspace(
    df["bm_ekonomi"].min(),
    df["bm_ekonomi"].max(),
    40
)

X1, X2 = np.meshgrid(x1, x2)

# MENGHITUNG NILAI Y PADA BIDANG REGRESI
Y = (
    intercept
    + coef_pendidikan * X1
    + coef_ekonomi * X2
)

# MEMBUAT FIGURE 3D
fig = go.Figure()

# MENAMPILKAN DATA AKTUAL
fig.add_trace(
    go.Scatter3d(
        x=df["bm_pendidikan"],
        y=df["bm_ekonomi"],
        z=df["KEMISKINAN"],
        mode="markers",
        marker=dict(
            size=7,
            color="red",
            opacity=0.9
        ),
        name="Data Aktual",
        text=[
            f"Tahun {t}"
            for t in df["Tahun"]
        ],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Belanja Modal Pendidikan: %{x:.2f}<br>"
            "Belanja Modal Ekonomi: %{y:.2f}<br>"
            "Kemiskinan: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

# MENAMPILKAN BIDANG REGRESI
fig.add_trace(
    go.Surface(
        x=X1,
        y=X2,
        z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Bidang Regresi",
        hovertemplate=(
            "Belanja Modal Pendidikan: %{x:.2f}<br>"
            "Belanja Modal Ekonomi: %{y:.2f}<br>"
            "Prediksi Kemiskinan: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

# MENGATUR TAMPILAN GRAFIK
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D - KOTA PEKANBARU"
            "<br>"
            f"<sup>{persamaan} | R² = {r_squared:.3f}</sup>"
        ),
        "x": 0.5
    },
    scene={
        "xaxis": {
            "title": "Belanja Modal Pendidikan (X1)"
        },
        "yaxis": {
            "title": "Belanja Modal Ekonomi (X2)"
        },
        "zaxis": {
            "title": "Kemiskinan (Y)"
        }
    },
    width=1000,
    height=750,
    margin={
        "l": 0,
        "r": 0,
        "b": 0,
        "t": 100
    },
    legend={
        "x": 0.02,
        "y": 0.98
    }
)

# MENAMPILKAN GRAFIK
fig.show()
```
